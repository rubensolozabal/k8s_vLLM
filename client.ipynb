```python
import os

import mlflow
from langchain.chat_models import init_chat_model
from langchain.messages import HumanMessage, SystemMessage

NODE_IP = "10.0.0.87"
API_URL = f"http://{NODE_IP}:31583/v1"
TRACKING_URL = f"http://localhost:5000"
API_KEY = "MLEXPERT_RULES"
MODEL = "Qwen/Qwen3-4B-Instruct-2507"

os.environ["MLFLOW_TRACKING_URI"] = TRACKING_URL
mlflow.langchain.autolog()
```

```python
model = init_chat_model(
    model=MODEL,
    model_provider="openai",
    base_url=API_URL,
    api_key=API_KEY,
    timeout=30,
    max_retries=0,
    max_completion_tokens=256,
)
```

## Conversation


```python
mlflow.set_experiment("conversation")
```

```python
system_msg = SystemMessage("You are an expert LLMOps engineer.")
human_msg = HumanMessage(
    "What are the top 3 best practices for LLMOps? Summarize each in a few words."
)

messages = [system_msg, human_msg]
```

```python
response = model.invoke(messages)
print(response.content)
```

```python
messages.append(response)
messages.append(
    HumanMessage(
        "So what are the top 3 tools for LLMOps? Summarize each in a few words."
    )
)
```

```python
%%time
response = model.invoke(messages)
print(response.content)
```

## Chat


```python
mlflow.set_experiment("chat")
```

```python
system_msg = SystemMessage(content="You are an expert LLMOps engineer.")
messages = [system_msg]

print("Chat with the expert LLMOps engineer! Type 'quit' to exit.")

while True:
    # Get user input from the notebook cell
    user_input = input("You: ")

    if user_input.lower() == "quit":
        print("Session ended.")
        break

    # Append the user's message to the history
    messages.append(HumanMessage(content=user_input))

    try:
        # Invoke the model with the entire message history
        response = model.invoke(messages)

        # Append the model's response to the history
        messages.append(response)

        # Print the model's response
        print(f"AI: {response.content}")

    except Exception as e:
        print(f"An error occurred: {e}")
        # Remove the last user message to allow for a retry
        messages.pop()
```

## Streaming


```python
mlflow.set_experiment("streaming")
```

```python
system_msg = SystemMessage("You are an expert LLMOps engineer.")
human_msg = HumanMessage("What are the top 3 best practices for LLMOps?")

messages = [system_msg, human_msg]

for chunk in model.stream(messages):
    print(chunk.content, end="", flush=True)
```
